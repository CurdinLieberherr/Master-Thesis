```python
START = 2000
END = 2024

COUNTRIES = [
    'Belgium',
    'Italy',
    'Portugal',
    'Sweden',
    'Spain'
]

allanalises = []

for country in COUNTRIES:
    if country == 'Portugal':
            start = 2007
    else: start = START
    analysis = MisallocationAnalysis(country, start, END, sector='C')
    allanalises.append(analysis)


for country in allanalises:
    print(country.country)
    fig = country.plot_capital_wedges_results()
    print('\n')
```

```python
for country in allanalises:
    print(country.country)
    df = country.estimate_capital_wedges()
    print(df.pivot(index='decile', columns='window',values=['tau_k_gap', 'se']).round(2))
    print('\n')
```
